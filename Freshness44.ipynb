```python
import os
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import load_model
from google.colab import drive
import pandas as pd
import kagglehub
import pickle
import json
import glob
import re

# Mount Drive
drive.mount('/content/drive')

# Project path
project_path = "/content/drive/MyDrive/FRESCO_project"
os.makedirs(project_path, exist_ok=True)

print("=" * 60)
print("FRESCO TRAINING WITH RESUME CAPABILITY")
print("=" * 60)

# 1. Get dataset path
print("\n📁 Loading dataset...")
dataset_path = kagglehub.dataset_download("siavash93/freshness44")
print(f"Dataset path: {dataset_path}")

# 2. Load splits
train_df = pd.read_csv(os.path.join(project_path, 'train.csv'))
val_df = pd.read_csv(os.path.join(project_path, 'val.csv'))
test_df = pd.read_csv(os.path.join(project_path, 'test.csv'))

# 3. Fix image paths
print("\n🔄 Fixing image paths...")
for df in [train_df, val_df, test_df]:
    df['image_path'] = df['image_path'].str.replace(
        '/root/.cache/kagglehub/datasets/siavash93/freshness44/versions/1',
        dataset_path,
        regex=False
    )

print(f"Training: {len(train_df)} images")
print(f"Validation: {len(val_df)} images")
print(f"Test: {len(test_df)} images")

# 4. Setup generators
print("\n📊 Creating data generators...")
IMG_SIZE = 224
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True,
    zoom_range=0.1,
    brightness_range=[0.8, 1.2],
    fill_mode='nearest'
)

val_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

def create_generator(df, datagen, batch_size=BATCH_SIZE):
    return datagen.flow_from_dataframe(
        dataframe=df,
        x_col='image_path',
        y_col='class_name',
        target_size=(IMG_SIZE, IMG_SIZE),
        batch_size=batch_size,
        class_mode='categorical',
        shuffle=True if df is train_df else False
    )

train_generator = create_generator(train_df, train_datagen)
val_generator = create_generator(val_df, val_datagen)
test_generator = create_generator(test_df, test_datagen)

steps_per_epoch = len(train_generator)
validation_steps = len(val_generator)

print(f"Steps per epoch: {steps_per_epoch}")
print(f"Validation steps: {validation_steps}")
print(f"Number of classes: {len(train_generator.class_indices)}")

# 5. Save class mapping
class_mapping = train_generator.class_indices
with open(os.path.join(project_path, 'class_mapping.json'), 'w') as f:
    json.dump(class_mapping, f)
print(f"✅ Class mapping saved ({len(class_mapping)} classes)")

# ============================================================
# 6. CHECK FOR EXISTING CHECKPOINT AND RESUME
# ============================================================

def get_latest_checkpoint():
    """Find the latest saved checkpoint"""
    checkpoint_dir = os.path.join(project_path, 'checkpoints')
    if not os.path.exists(checkpoint_dir):
        return None, 0

    # Find all .keras files in checkpoint directory
    checkpoint_files = glob.glob(os.path.join(checkpoint_dir, 'epoch_*.keras'))

    if not checkpoint_files:
        return None, 0

    # Extract epoch numbers
    epoch_numbers = []
    for f in checkpoint_files:
        match = re.search(r'epoch_(\d+)\.keras', f)
        if match:
            epoch_numbers.append(int(match.group(1)))

    if not epoch_numbers:
        return None, 0

    latest_epoch = max(epoch_numbers)
    latest_file = os.path.join(checkpoint_dir, f'epoch_{latest_epoch}.keras')
    return latest_file, latest_epoch

# Check for existing checkpoint
checkpoint_file, last_epoch = get_latest_checkpoint()

# ============================================================
# 7. BUILD OR LOAD MODEL
# ============================================================

if checkpoint_file and os.path.exists(checkpoint_file):
    print(f"\n✅ Found checkpoint: {checkpoint_file}")
    print(f"   Resuming from epoch {last_epoch}")
    model = load_model(checkpoint_file)
    print("✅ Model loaded successfully!")
else:
    print("\n🏗️ No checkpoint found. Building new model...")
    base_model = MobileNetV2(
        input_shape=(IMG_SIZE, IMG_SIZE, 3),
        include_top=False,
        weights='imagenet'
    )
    base_model.trainable = False

    model = keras.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(128, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(len(train_generator.class_indices), activation='softmax')
    ])

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.001),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    last_epoch = 0
    print("✅ New model built!")

model.summary()

# ============================================================
# 8. TRAIN WITH AUTO-SAVE AFTER EVERY EPOCH
# ============================================================

from tensorflow.keras.callbacks import Callback

class AutoSaveCheckpoint(Callback):
    """Save model after every epoch to Google Drive"""
    def __init__(self, save_dir, start_epoch=0):
        super().__init__()
        self.save_dir = save_dir
        self.start_epoch = start_epoch
        os.makedirs(save_dir, exist_ok=True)

    def on_epoch_end(self, epoch, logs=None):
        epoch_number = self.start_epoch + epoch + 1
        save_path = os.path.join(self.save_dir, f'epoch_{epoch_number}.keras')
        self.model.save(save_path)
        print(f"\n💾 Checkpoint saved: epoch_{epoch_number}.keras")

# Create checkpoint directory
checkpoint_dir = os.path.join(project_path, 'checkpoints')
os.makedirs(checkpoint_dir, exist_ok=True)

# ============================================================
# 9. PHASE 1: FROZEN BASE (10 EPOCHS)
# ============================================================

print("\n" + "=" * 60)
print("PHASE 1: Training with frozen base")
print(f"Resuming from epoch {last_epoch}")
print("=" * 60)

# Determine remaining epochs
phase1_total_epochs = 10
remaining_epochs = phase1_total_epochs - last_epoch

if remaining_epochs > 0:
    # Create checkpoint callback
    checkpoint_callback = AutoSaveCheckpoint(
        save_dir=checkpoint_dir,
        start_epoch=last_epoch
    )

    # Train
    history_phase1 = model.fit(
        train_generator,
        steps_per_epoch=steps_per_epoch,
        epochs=phase1_total_epochs,
        initial_epoch=last_epoch,
        validation_data=val_generator,
        validation_steps=validation_steps,
        callbacks=[checkpoint_callback],
        verbose=1
    )

    # Save final phase 1 model
    model.save(os.path.join(project_path, 'fresco_model_phase1.h5'))
    print("✅ Phase 1 complete!")
else:
    print("✅ Phase 1 already complete. Loading saved model...")
    model = load_model(os.path.join(project_path, 'fresco_model_phase1.h5'))

# ============================================================
# 10. PHASE 2: FINE-TUNING (15 EPOCHS)
# ============================================================

print("\n" + "=" * 60)
print("PHASE 2: Fine-tuning top 30 layers")
print("=" * 60)

# Check if phase 2 is already done
phase2_checkpoint = os.path.join(checkpoint_dir, 'phase2_complete.txt')

if not os.path.exists(phase2_checkpoint):
    # Unfreeze top 30 layers
    base_model = model.layers[0]
    base_model.trainable = True
    for layer in base_model.layers[:-30]:
        layer.trainable = False
    for layer in base_model.layers[-30:]:
        layer.trainable = True

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-5),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    # Check for phase 2 checkpoint
    phase2_last_epoch = 0
    phase2_files = glob.glob(os.path.join(checkpoint_dir, 'phase2_epoch_*.keras'))
    if phase2_files:
        epoch_numbers = []
        for f in phase2_files:
            match = re.search(r'phase2_epoch_(\d+)\.keras', f)
            if match:
                epoch_numbers.append(int(match.group(1)))
        if epoch_numbers:
            phase2_last_epoch = max(epoch_numbers)
            latest_phase2 = os.path.join(checkpoint_dir, f'phase2_epoch_{phase2_last_epoch}.keras')
            print(f"✅ Found Phase 2 checkpoint: epoch {phase2_last_epoch}")
            model = load_model(latest_phase2)

    # Create phase 2 checkpoint callback
    class Phase2Checkpoint(Callback):
        def __init__(self, save_dir, start_epoch=0):
            super().__init__()
            self.save_dir = save_dir
            self.start_epoch = start_epoch
            os.makedirs(save_dir, exist_ok=True)

        def on_epoch_end(self, epoch, logs=None):
            epoch_number = self.start_epoch + epoch + 1
            save_path = os.path.join(self.save_dir, f'phase2_epoch_{epoch_number}.keras')
            self.model.save(save_path)
            print(f"\n💾 Phase 2 checkpoint saved: phase2_epoch_{epoch_number}.keras")

    phase2_callback = Phase2Checkpoint(
        save_dir=checkpoint_dir,
        start_epoch=phase2_last_epoch
    )

    # Train Phase 2
    phase2_total_epochs = 15
    remaining_phase2 = phase2_total_epochs - phase2_last_epoch

    if remaining_phase2 > 0:
        history_phase2 = model.fit(
            train_generator,
            steps_per_epoch=steps_per_epoch,
            epochs=phase2_total_epochs,
            initial_epoch=phase2_last_epoch,
            validation_data=val_generator,
            validation_steps=validation_steps,
            callbacks=[phase2_callback],
            verbose=1
        )

        # Mark Phase 2 complete
        with open(phase2_checkpoint, 'w') as f:
            f.write(f"Phase 2 complete at epoch {phase2_total_epochs}\n")
            f.write(f"Test accuracy will be below\n")

    # Save final model
    model.save(os.path.join(project_path, 'fresco_model_final.h5'))
    model.save(os.path.join(project_path, 'fresco_model_final.keras'))
    print("✅ Phase 2 complete! Final model saved.")
else:
    print("✅ Phase 2 already complete.")
    model = load_model(os.path.join(project_path, 'fresco_model_final.h5'))

# ============================================================
# 11. EVALUATE ON TEST SET
# ============================================================

print("\n" + "=" * 60)
print("📊 EVALUATING ON TEST SET")
print("=" * 60)

test_loss, test_accuracy = model.evaluate(test_generator, verbose=1)
print(f"\nTest Accuracy: {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")
print(f"Test Loss: {test_loss:.4f}")

# Save results
with open(os.path.join(project_path, 'test_results.txt'), 'w') as f:
    f.write("=" * 60 + "\n")
    f.write("FRESCO MODEL TEST RESULTS\n")
    f.write("=" * 60 + "\n\n")
    f.write(f"Test Accuracy: {test_accuracy:.4f} ({test_accuracy*100:.2f}%)\n")
    f.write(f"Test Loss: {test_loss:.4f}\n")
    f.write(f"Number of classes: {len(train_generator.class_indices)}\n")
    f.write(f"Training images: {len(train_df)}\n")
    f.write(f"Validation images: {len(val_df)}\n")
    f.write(f"Test images: {len(test_df)}\n")

print("✅ Results saved!")

# ============================================================
# 12. SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("✅ TRAINING COMPLETE!")
print("=" * 60)
print(f"Test Accuracy: {test_accuracy:.4f} ({test_accuracy*100:.2f}%)")
print(f"Classes: {len(train_generator.class_indices)}")
print(f"\n📁 Files saved to: {project_path}")
print("📁 Checkpoints saved to: checkpoints/")
print("=" * 60)
```

```python
import os
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from google.colab import drive
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json
import kagglehub # Required to re-download or confirm dataset location

# Mount Google Drive (if not already mounted)
drive.mount('/content/drive')

# Project path (where metadata and model will be saved)
project_path = "/content/drive/MyDrive/FRESCO_project"
os.makedirs(project_path, exist_ok=True)

# 1. Ensure the dataset is downloaded and get its current path
print("Ensuring dataset is downloaded to get correct image root path...")
current_kaggle_download_path = kagglehub.dataset_download("siavash93/freshness44")
print(f"Dataset root is currently at: {current_kaggle_download_path}")

# 2. Load the splits metadata from Google Drive
print("\nLoading split dataframes from Google Drive...")
train_df = pd.read_csv(os.path.join(project_path, 'train.csv'))
val_df = pd.read_csv(os.path.join(project_path, 'val.csv'))
test_df = pd.read_csv(os.path.join(project_path, 'test.csv'))

# 3. Correct the 'image_path' column in the dataframes
# The paths stored in the CSVs have an old base, e.g., '/root/.cache/kagglehub/.../versions/1/Freshness44/...'
# We need to replace the *base directory part* with the *new current root path* from kagglehub.dataset_download.

# Get a sample image path from the loaded dataframe to extract its old base.
sample_image_path_from_df = train_df['image_path'].iloc[0]

# Dynamically determine the 'old_base_path' from the sample_image_path_from_df
# This assumes the dataset's top-level folder is 'Freshness44' and it's directly under the base path.
# We find the part of the string that ends right before 'Freshness44'.
freshness44_start_index = sample_image_path_from_df.find('Freshness44')
if freshness44_start_index != -1:
    # old_base_path_in_df will be something like '/root/.cache/kagglehub/datasets/siavash93/freshness44/versions/1'
    old_base_path_in_df = sample_image_path_from_df[:freshness44_start_index - 1] # -1 to remove the '/' before Freshness44

    # Now replace this dynamically found old base path with the new current_kaggle_download_path
    train_df['image_path'] = train_df['image_path'].str.replace(old_base_path_in_df, current_kaggle_download_path, regex=False)
    val_df['image_path'] = val_df['image_path'].str.replace(old_base_path_in_df, current_kaggle_download_path, regex=False)
    test_df['image_path'] = test_df['image_path'].str.replace(old_base_path_in_df, current_kaggle_download_path, regex=False)
    print("Image paths updated successfully in dataframes.")
else:
    print("Warning: 'Freshness44' not found in image_path from dataframe. Paths might not be updated correctly.")
    print("Assuming image paths are already correct or relative, proceeding without modification.")

print(f"Updated Training images: {len(train_df)}")
print(f"Updated Validation images: {len(val_df)}")
print(f"Updated Test images: {len(test_df)}")
print("\nSample of updated image paths:")
print(train_df['image_path'].head())

# 4. Define image size and batch size
IMG_SIZE = 224
BATCH_SIZE = 32

# 5. Data augmentation and generators
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True,
    zoom_range=0.1,
    brightness_range=[0.8, 1.2],
    fill_mode='nearest'
)

val_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

def create_generator(df, datagen, batch_size=BATCH_SIZE):
    return datagen.flow_from_dataframe(
        dataframe=df,
        x_col='image_path',
        y_col='class_name',
        target_size=(IMG_SIZE, IMG_SIZE),
        batch_size=batch_size,
        class_mode='categorical',
        shuffle=True if df is train_df else False # Shuffle only training data
    )

print("\nCreating image data generators...")
train_generator = create_generator(train_df, train_datagen)
val_generator = create_generator(val_df, val_datagen)
test_generator = create_generator(test_df, test_datagen)

# 6. Load class mapping
class_mapping_path = os.path.join(project_path, 'class_mapping.json')
class_mapping = {} # Initialize as empty

if os.path.exists(class_mapping_path):
    try:
        with open(class_mapping_path, 'r') as f:
            class_mapping = json.load(f)
        if not class_mapping: # If file existed but was empty JSON
            print(f"Warning: {class_mapping_path} exists but is empty or invalid. Re-creating from train_generator.")
            class_mapping = train_generator.class_indices
            if class_mapping: # Only save if not empty
                with open(class_mapping_path, 'w') as f:
                    json.dump(class_mapping, f)
                print(f"Recreated and saved class mapping with {len(class_mapping)} classes.")
            else:
                print("Re-creation from train_generator also resulted in empty class mapping. Check data and generator setup.")
        else:
            print(f"\nLoaded class mapping with {len(class_mapping)} classes.")
    except json.JSONDecodeError: # If file was corrupted or non-JSON
        print(f"Error decoding {class_mapping_path}. Re-creating from train_generator.")
        class_mapping = train_generator.class_indices
        if class_mapping: # Only save if not empty
            with open(class_mapping_path, 'w') as f:
                json.dump(class_mapping, f)
            print(f"Recreated and saved class mapping with {len(class_mapping)} classes.")
        else:
            print("Re-creation from train_generator also resulted in empty class mapping. Check data and generator setup.")
else:
    print("\nclass_mapping.json not found! Creating from train_generator.")
    class_mapping = train_generator.class_indices
    if class_mapping: # Only save if not empty
        with open(class_mapping_path, 'w') as f:
            json.dump(class_mapping, f)
        print(f"Created and saved class mapping with {len(class_mapping)} classes.")
    else:
        print("Creation from train_generator also resulted in empty class mapping. Check data and generator setup.")

# Ensure class_mapping is not empty before proceeding to model creation
if not class_mapping:
    raise ValueError("Class mapping is empty. Ensure your dataframes and generators are correctly configured and contain valid image paths and class names.")


print("\nSample class mapping (first 10 items):")
for i, (cls, idx) in enumerate(list(class_mapping.items())[:10]):
    print(f"  {cls}: {idx}")

# 7. Re-create the model architecture
print("\nRe-creating MobileNetV2 model architecture...")
base_model = MobileNetV2(
    input_shape=(IMG_SIZE, IMG_SIZE, 3),
    include_top=False,
    weights='imagenet'
)
base_model.trainable = False

model = keras.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(len(class_mapping), activation='softmax')
])

# 8. Compile the model
model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("\nModel Summary:")
model.summary()

# 9. Check if a previously trained model (phase 1) exists and load its weights
model_phase1_path = os.path.join(project_path, 'fresco_model_phase1.h5')
if os.path.exists(model_phase1_path):
    print(f"\nLoading weights from previously saved model: {model_phase1_path}")
    model.load_weights(model_phase1_path)
    print("✅ Model weights loaded successfully!")
    print("However, based on previous execution logs, the model was not found, so this might be a fresh start.")
    print("We will re-train for the full 10 epochs as the previous training did not complete.")
else:
    print("\n❌ No previously saved model (fresco_model_phase1.h5) found for phase 1. Training will start from scratch.")

# Always start from epoch 0 if no saved model or previous training didn't complete/save.
initial_epoch_to_resume = 0

# 10. Calculate steps per epoch
steps_per_epoch = train_generator.samples // BATCH_SIZE
validation_steps = val_generator.samples // BATCH_SIZE

print(f"\nSteps per epoch: {steps_per_epoch}")
print(f"Validation steps: {validation_steps}")

print(f"\nStarting training for 10 epochs...")
history_phase1 = model.fit(
    train_generator,
    steps_per_epoch=steps_per_epoch,
    epochs=10,
    validation_data=val_generator,
    validation_steps=validation_steps,
    verbose=1
)

# 11. Save the model after phase 1 training
model.save(os.path.join(project_path, 'fresco_model_phase1.h5'))
print(f"\n✅ Model saved to: {project_path}/fresco_model_phase1.h5")
```
